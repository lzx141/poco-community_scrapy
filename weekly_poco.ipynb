```python
def parse_days_ago(text):
    """将相对时间转换为 datetime 对象"""
    if not text:
        return None
    
    now = datetime.datetime.now()
    text = str(text).strip()
    
    try:
        # 处理各种时间格式
        match = re.match(r'(\d+)\s+Day[s]?\s+ago', text, re.IGNORECASE)
        if match:
            return now - datetime.timedelta(days=int(match.group(1)))
        
        match = re.match(r'(\d+)\s+Hour[s]?\s+ago', text, re.IGNORECASE)
        if match:
            return now - datetime.timedelta(hours=int(match.group(1)))
        
        match = re.match(r'(\d+)\s+Minute[s]?\s+ago', text, re.IGNORECASE)
        if match:
            return now - datetime.timedelta(minutes=int(match.group(1)))
        
        if 'Just now' in text or 'Just Now' in text:
            return now
        
        if 'Yesterday' in text or 'yesterday' in text:
            return now - datetime.timedelta(days=1)
        
        # 如果已经是完整日期，尝试解析
        try:
            return date_parse(text)
        except:
            return text
            
    except Exception:
        return text

for i in range(len(df)):
    print(f'抓取第{i+1}个链接')
    url = f'https://community.poco.in/post/{df.loc[i, "id"]}'
    
    try:
        # 使用driver访问页面
        driver.get(url)
        time.sleep(random.uniform(12, 16))
        
        # 等待页面加载
        try:
            WebDriverWait(driver, 20).until(
                EC.presence_of_element_located((By.XPATH, '//*[@id="root"]/div[2]/div[2]/div/div[1]/div[1]'))
            )
        except:
            print(f"页面加载超时: {url}")
            continue
        
        # 获取页面源码并解析
        res = driver.page_source
        data = etree.HTML(res)
        
        # 提取标题和内容
        title = data.xpath('string(//*[@id="root"]/div[2]/div[2]/div/div[1]/div[1]/div[1]/span/strong)')
        content = data.xpath('string(//*[@id="root"]/div[2]/div[2]/div/div[1]/div[1]/div[2]/div)')
        
        df.loc[i, 'url'] = url
        df.loc[i, 'title'] = title
        df.loc[i, 'content'] = content
        df.loc[i, 'isParent'] = 1
        
        print(f"标题长度: {len(title)}, 内容长度: {len(content)}")
    
        # 提取评论
        comment_sections = data.xpath('//div[contains(@class, "comment-section")]//div[contains(@class, "MuiListItem-root")]')
        
        if comment_sections:
            comments_list = []  # 使用不同的变量名
            for item in comment_sections:
                newdata = {}
                
                # 获取评论者用户名
                username_elem = item.xpath('.//a[contains(@href, "/user/")]')
                if username_elem:
                    newdata['postUserName'] = username_elem[0].text.strip() if username_elem[0].text else ""
                    authorURL = username_elem[0].get('href', '')
                    newdata['userId'] = authorURL.split('/')[-1] if authorURL else ""
                else:
                    newdata['postUserName'] = ""
                    newdata['userId'] = ""
                
                # 获取评论时间
                time_elem = item.xpath('.//span[contains(@class, "MuiTypography-caption")]')
                if time_elem:
                    publishDate = time_elem[0].text.strip() if time_elem[0].text else ""
                    # 处理时间
                    time_obj = parse_days_ago(publishDate)
                    if isinstance(time_obj, datetime.datetime):
                        newdata['publishDate'] = time_obj.strftime("%Y-%m-%d")
                    elif publishDate:
                        try:
                            newdata['publishDate'] = date_parse(publishDate).strftime("%Y-%m-%d")
                        except:
                            newdata['publishDate'] = publishDate
                    else:
                        newdata['publishDate'] = ""
                else:
                    newdata['publishDate'] = ""
                
                # 获取评论内容
                info_list = data.xpath('//div/span[contains(@class, "comment-section")]/span[contains(@class, "MuiList-root")]')
                if len(info_list):
                    data_list = []
                    for item in info_list:
                        newdata = {}
                        newdata['postUserName'] = item.xpath('string(./span//div[contains(@class, "MuiListItemText-root")]/span/a)').strip()
                        authorURL = item.xpath('string(./span//div[contains(@class, "MuiListItemText-root")]/span/a/@href)')
                        newdata['userId'] = authorURL.split('/')[-1] if authorURL else ""
                
                        publishDate = item.xpath('string(./span//div[contains(@class, "MuiListItemText-root")]/span/span)')
                        if publishDate:
                            time_obj = parse_days_ago(publishDate)
                            if isinstance(time_obj, datetime.datetime):
                                newdata['publishDate'] = time_obj.strftime("%Y-%m-%d")
                            else:
                                try:
                                    newdata['publishDate'] = date_parse(str(publishDate)).strftime("%Y-%m-%d")
                                except:
                                    newdata['publishDate'] = str(publishDate)
                        else:
                            newdata['publishDate'] = ""
                
                            newdata['content'] = item.xpath('string(./span//div[contains(@class, "MuiListItemText-root")]/div)')
                            newdata['isParent'] = 0
                            newdata['prUrl'] = url
                
                        if not len(newdata['content']):
                            print('评论没有内容')
                
                        data_list.append(newdata)
                        data_list_all.append(newdata)
            
                    print(f'共有{len(data_list)}条回复')
                
    
    except Exception as e:
        print(f"处理链接 {url} 时出错: {e}")
        import traceback
        traceback.print_exc()  # 打印完整错误堆栈
        continue
```
